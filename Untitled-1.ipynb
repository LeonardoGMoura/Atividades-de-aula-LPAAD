### O que é uma classe ?

Uma classe é estrutura que define caracteristicas e comportamentos que seus objetos poderão pssuir 

Função classe:
Definir quais informações e quais comportamentos serão comuns a todos os objetos criados a partir dela

Exemplo:
A classe musica representa uma estrutura comum para diferentes musicas.Ela pode estabelecer que as musicas possuam titulo,artista,duração e operações como registar reproduções

O que a classe representa: Um modelo(O molde) que define os atributos e os metodos que todos os objetos criados a partir dela deverão possuir



### O que é um objeto

Um objeto é uma entidade criada apartir de uma classe que possui identidades propria e pode armazenar informações e executar

### O que é _init_

O init é um metodo especial do python executado automaticamento durante a inicialização de uma instancia a´pos a sua criação. Ele perimte estabelecer os valores iniciais dos atributos de cada objeto

```python
class Musica():
    def __init__(self, nome, cantor, duracao):
        self.nome = nome
        self.cantor = cantor
        self.duracao = duracao
        
m1 = Musica("Vida Boa", "Victor e Leo", 3.5)
```

### Entenda o que é self
self é o parametro convencional para referenciar a propria instacia dentro de uma mesma instancia

### O que são métodos
métodos são funções definidas dentro de uma classe que representam operações ou comportamentos associados a seus objetos

exemplo
uma musica pode ser executar as seguintes operações:

exibir as suas informações 

registar uma reprodução

atualizar a sua avaliação

```python
def exibir_informacoes(nome,cantor,duracao):
    print(f"Nome: {nome}")
    print(f"Cantor: {cantor}")
    print(f"Duração: {duracao} minutos")

exibir_informacoes(m1.nome, m1.cantor, m1.duracao)
```

| Atributos | Tipo |
|-----------|----- |
| nome      | str  |
| cantor    | str  |
| genero    | str  |
| duracao   | int  |
| ano       | int  |

```python
class Musica:
    def __init__(self, nome, cantor,genero, duracao,ano):
        self.nome = nome
        self.cantor = cantor
        self.genero = genero
        self.duracao = duracao
        self.ano = ano

    # atribuindoreproduzidos
        self.reproducoes = 0
        self.tocando = False,bool # é uma boleana


    # padronize os nomes das funções
    #evite colocar o nome da função, com o mesmo nome do atributo

    def mostar_nome(self):
        print(f"Nome: {self.nome}")
    def mostar_cantor(self):
        print(f"Cantor: {self.cantor}")
    def mostar_genero(self):
        print(f"Gênero: {self.genero}")
    def mostar_duracao(self):
        print(f"Duração: {self.duracao} minutos")
    def mostar_ano(self):
        print(f"Ano: {self.ano}")

    # métod de exibir todas as informações base da música
    def mostar_detalhe(self):
        self.mostar_nome()
        self.mostar_cantor()
        self.mostar_genero()
        self.mostar_duracao()
        self.mostar_ano()
    #----------------------------------------------------
    def reproduzir(self):
        if self.tocando == False:
            self.tocando = True
            self.reproducoes += 1
            print(f"reproduzindo {self.nome}")
        else:
            print(f"Essa musica ja esta sendo executada")

    def pausar(self):
        if self.tocando == True:
            self.tocando = False
            print(f"musica pausada!")
        else:
            print(f"Nenhuma musica está sendo reproduzida")
```

```python
m1 = Musica("Caneta Azul","Manoel Gomes","Brega",73,2019)

m1.mostar_nome()
m1.mostar_cantor()
m1.mostar_genero()
m1.mostar_duracao()
m1.mostar_ano()


print("\ninformações completas:")

m1.mostar_detalhe()


```

```python
m1.reproduzir()
print(m1.reproducoes)

m1.reproduzir()

m1.pausar()

m1.reproduzir()
print(m1.reproducoes)
```
