Parte 1: Processamento dos chamados

Desenvolva um programa em Python que:
1. Percorra a lista de chamados utilizando enumerate(), numerando os registros a partir de 1;
2. Converta as datas de abertura e fechamento, atualmente armazenadas como strings, para
objetos datetime;

3. Calcule o tempo de atendimento, em dias, de cada chamado;

4. Apresente, para cada chamado:
• número do registro;
• código;
• responsável;
• prioridade;
• tempo de atendimento em dias;
• situação do prazo.

5. Classifique cada chamado como:
• Dentro do prazo: tempo de atendimento menor ou igual a 3 dias;
• Fora do prazo: tempo de atendimento maior que 3 dias.
Uma possível apresentação para cada registro é:

1 - CH001 | Maria | Alta | 2 dias | Dentro do prazo

```python
from datetime import datetime
```

```python
chamados = [
{
"codigo": "CH001",
"responsavel": "Maria",
"abertura": "15/09/2026",
"fechamento": "17/09/2026",
"prioridade": "Alta"
},

{
"codigo": "CH002",
"responsavel": "Carlos",
"abertura": "16/09/2026",
"fechamento": "22/09/2026",
"prioridade": "Média"
},

{
"codigo": "CH003",
"responsavel": "Marina",
"abertura": "18/09/2026",
"fechamento": "19/09/2026",
"prioridade": "Alta"
},

{
"codigo": "CH004",
"responsavel": "Rafael",
"abertura": "20/09/2026",
"fechamento": "25/09/2026",
"prioridade": "Baixa"
},

{
"codigo": "CH005",
"responsavel": "Maria",
"abertura": "21/09/2026",
"fechamento": "23/09/2026",
"prioridade": "Média"
}

]
```

```python
for chamado in chamados:
    for chave,valor in chamado.items():

        if chave == "abertura" or chave == "fechamento":

            valor = datetime.strptime(valor,"%d/%m/%Y")
            print(type(valor))
        print(f"{chave}: {valor}")
```

```python
for numero, chamado in enumerate(chamados, start=1):
    abertura = datetime.strptime(chamado["abertura"], "%d/%m/%Y")
    fechamento = datetime.strptime(chamado["fechamento"], "%d/%m/%Y")
    tempo = (fechamento - abertura).days
    print(f"{numero} - {chamado['codigo']}  Tempo de atendimento: {tempo} dias")
```

```python
fora_do_prazo = []

dentro_prazo = 0
fora = 0
maior_tempo = 0
codigo_maior = ""


chamados_prioridade = {
    "Alta": [],
    "Média": [],
    "Baixa": []
}

for numero, chamado in enumerate(chamados, start=1):
    abertura = datetime.strptime(chamado["abertura"], "%d/%m/%Y")
    fechamento = datetime.strptime(chamado["fechamento"], "%d/%m/%Y")
    tempo = (fechamento - abertura).days

    if tempo <= 3:
        situacao = "Dentro do prazo"
        dentro_prazo += 1
    else:
        situacao = "Fora do prazo"
        fora += 1
        fora_do_prazo.append({
            "codigo": chamado["codigo"],
            "responsavel": chamado["responsavel"],
            "tempo": tempo
        })

    if tempo > maior_tempo:
        maior_tempo = tempo
        codigo_maior = chamado["codigo"]

    chamados_prioridade[chamado["prioridade"]].append(chamado["codigo"])

    print(f"{numero} - {chamado['codigo']}  {chamado['responsavel']}  {chamado['prioridade']} {tempo} dias  {situacao}")
```
