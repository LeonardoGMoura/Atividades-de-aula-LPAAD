Atividade Prática 2: Análise de Transações Bancárias

Parte 1: Processamento das transações

```python
from datetime import datetime, timedelta
import statistics
```

```python
transacoes = [
    {"codigo": "TR001", "cliente": "Maria", "tipo": "PIX", "valor": 1250.00, "horario": "08:15", "processamento": 2},
    {"codigo": "TR002", "cliente": "Carlos", "tipo": "Transferência", "valor": 3500.00, "horario": "09:40", "processamento": 8},
    {"codigo": "TR003", "cliente": "Marina", "tipo": "Pagamento", "valor": 850.00, "horario": "10:25", "processamento": 5},
    {"codigo": "TR004", "cliente": "Rafael", "tipo": "PIX", "valor": 1250.00, "horario": "11:50", "processamento": 3},
    {"codigo": "TR005", "cliente": "Beatriz", "tipo": "TED", "valor": 4200.00, "horario": "13:35", "processamento": 15},
    {"codigo": "TR006", "cliente": "Lucas", "tipo": "PIX", "valor": 2100.00, "horario": "14:20", "processamento": 2},
    {"codigo": "TR007", "cliente": "Fernanda", "tipo": "Pagamento", "valor": 850.00, "horario": "15:45", "processamento": 6}
]
```

```python
for numero, transacao in enumerate(transacoes, start=1):
    print(f"{numero} - {transacao}")
```

```python
for numero, transacao in enumerate(transacoes, start=1):
    inicio = datetime.strptime(transacao["horario"], "%H:%M")
    print(f"{numero} - {transacao['codigo']}  {inicio}")
```

```python
valores = []
acima_da_media = [] 

for numero, transacao in enumerate(transacoes, start=1):
    inicio = datetime.strptime(transacao["horario"], "%H:%M")
    conclusao = inicio + timedelta(minutes=transacao["processamento"])
    valores.append(transacao["valor"])

    print(f"{numero} - {transacao['codigo']} {transacao['cliente']}  {transacao['tipo']}  "
          f"{transacao['valor']:.2f} reais  {inicio.strftime('%H:%M')} "
          f"Conclusão: {conclusao.strftime('%H:%M')}")
```

parte 2 : Análise estatística dos valores

```python
media = statistics.mean(valores)
mediana = statistics.median(valores)
moda = statistics.multimode(valores)
variancia = statistics.pvariance(valores)
desvio = statistics.pstdev(valores)

print("Média:", media)
print("Mediana:", mediana)
print("Moda:", moda)
print("Variância populacional:", variancia)
print("Desvio padrão populacional:", desvio)
print("Menor valor:", min(valores))
print("Maior valor:", max(valores))
```

Parte 3: Análise das transações

```python
qtd_acima = 0
qtd_abaixo = 0

for numero, transacao in enumerate(transacoes, start=1):
    if transacao["valor"] > media:
        classificacao = "Acima da média"
        qtd_acima += 1
        acima_da_media.append({
            "codigo": transacao["codigo"],
            "cliente": transacao["cliente"],
            "tipo": transacao["tipo"],
            "valor": transacao["valor"]
        })
    elif transacao["valor"] == media:
        classificacao = "Dentro da média"
    else:
        classificacao = "Abaixo da média"
        qtd_abaixo += 1

    print(f"{numero} - {transacao['codigo']} {transacao['cliente']}  "
          f"{transacao['valor']:.2f} reais  {classificacao}")
```

Parte 4: Resumo das transações

```python
print("Quantidade total de transações:", len(transacoes))
print(f"Valor total movimentado é: R$ {sum(valores)}")
print("Menor valor:", min(valores))
print("Maior valor:", max(valores))
print("Transações acima da média:", qtd_acima)
print("Transações abaixo da média:", qtd_abaixo)
```
