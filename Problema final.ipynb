### Projeto Final - Processamento de encomendas

Uma empresa de logística realiza o transporte de encomendas para diferentes regiões do
Brasil. Cada solicitação pode conter um ou mais pacotes, com pesos distintos, e o custo do
transporte depende da região de destino, da modalidade escolhida e da contratação de seguro.
Para automatizar esse processo, a empresa deseja desenvolver um sistema que valide as
informações recebidas, calcule o frete e determine se a encomenda pode ser aprovada

```python
def calcular_frete(cliente, *pesos, **informacoes):

    if not isinstance(cliente, str) or not cliente:
        print("cliente invalido")
        return

    for peso in pesos:
        if not isinstance(peso, (int, float)):
            print("o peso deve ser um numero inteiro ou decimal")
            return
        if peso <= 0 or peso > 30:
            print("peso invalido")
            return

    peso_total = sum(pesos)
    if peso_total > 100:
        print("o peso total ultrapassou o limite")
        return

    tarifas = {
        "Sudeste": 5,
        "Sul": 6,
        "Centro-Oeste": 7,
        "Nordeste": 8,
        "Norte": 10
    }

    if "regiao" not in informacoes or informacoes["regiao"] not in tarifas:
        print("regiao invalida")
        return
    tarifa_r = tarifas[informacoes["regiao"]]

    frete_base = peso_total * tarifa_r

    if "modalidade" in informacoes:
                 if informacoes["modalidade"] not in ["expressa","econômica"]:
                        print("modalidade invalida")

    if informacoes["modalidade"] == ["econômica"]:
        frete_expressa = frete_base
        print(f"não possui acréscimos sobre o frete base")
    else:
        frete_expressa = frete_base * 1.4
        print(f"possui acréscimo de 40% sobre o frete base")

    if "seguro" in informacoes:
        if not isinstance(informacoes["seguro"], bool):
            print("seguro invalido")
            return

    if "seguro" in informacoes and informacoes["seguro"] == True:
        seguro = frete_expressa * 0.05
        print("seguro aplicado")
    else:
        seguro = 0
        print("seguro não aplicado")

    frete_final = frete_expressa + seguro

    print(f"Cliente: {cliente}")
    print(f"Peso total: {peso_total} kg")
    print(f"Frete base: R$ {frete_base:.2f}")
    print(f"Frete com modalidade: R$ {frete_expressa:.2f}")
    print(f"Seguro: R$ {seguro:.2f}")
    print(f"Frete final: R$ {frete_final:.2f}")
    print("A encomenda esta aprovada e pronta para o envio")

    return frete_final
```
